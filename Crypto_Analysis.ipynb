# Cryptocurrency Data Analysis

This notebook replicates the workflow from the article *"Cryptocurrency Data Analysis Part I: Obtaining and Playing with Data of Digital Assets"* using modern Python tools and the Binance API. We will fetch historical data, visualize it, and perform basic statistical analysis to understand volatility.

```python
import pandas as pd
import matplotlib.pyplot as plt
from crypto_data import get_crypto_data

# Ensure plots appear inline
%matplotlib inline
```

## 1. Fetching the Data

We use our custom `get_crypto_data` function which pulls data from the Binance API. We'll fetch the daily close data for Bitcoin (`BTCUSDT`).

```python
# Fetch up to 1000 days of daily BTC data
df_full = get_crypto_data('BTCUSDT', interval='1d', limit=1000)

# We are primarily interested in the 'close' price for this analysis
df = df_full['close']
```

## 2. Getting the First Look

Let's inspect the head of the dataframe to ensure the dates and prices look correct.

```python
df.head()
```

## 3. Visualizing Historical Price

To visually inspect the historical time-series, we can simply call `.plot()`.

```python
plt.figure(figsize=(12, 6))
df.plot(title="BTC/USDT Historical Daily Close Price", color='orange')
plt.ylabel("Price (USDT)")
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Analyzing Returns and Volatility

Pandas makes it easy to analyze the percentage returns. Let's look at the descriptive statistics of the daily percentage returns.

```python
returns = df.pct_change().dropna()
returns.describe()
```

Bitcoin is known for its volatility and "fat tails". We can visualize this by plotting a histogram of the percentage returns.

```python
plt.figure(figsize=(10, 6))
returns.hist(bins=100, color='skyblue', edgecolor='black', alpha=0.7)
plt.title("BTC/USDT Daily Close Returns Distribution")
plt.xlabel("Daily Return")
plt.ylabel("Frequency")
plt.grid(True, alpha=0.3)
plt.show()
```
